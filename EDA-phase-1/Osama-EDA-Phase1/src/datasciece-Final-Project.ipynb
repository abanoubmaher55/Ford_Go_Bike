###  To fix this issue You will fill these values with mean for numeric and mode for categorical

```python
rat = preprocessing.Ratio(df)

num_col = df.select_dtypes("number").columns
cat_col = df.select_dtypes("category").columns

for i in rat.index:

    ratio = rat.loc[i, "Ratio"]

    if ratio > 50:
        df.drop(i, axis=1, inplace=True)

    elif 5 <= ratio <= 50:
        df.dropna(subset=[i], inplace=True)

    elif ratio < 5:
        if i in num_col:
            df[i] = df[i].fillna(df[i].mean())

        elif i in cat_col:
            df[i] = df[i].fillna(df[i].mode()[0])

preprocessing.Ratio(df)
```

```python
#check the Nulls 
df.isnull().sum()
```
