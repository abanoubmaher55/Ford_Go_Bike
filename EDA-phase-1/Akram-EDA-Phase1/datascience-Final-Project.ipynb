
# <a id="4">  4 ) . Do the users' ages have outlier values?</a>


> ### <mark>Your analsys and exploration here </mark>

- Presence of Outliers:
The distribution contains clear outliers on the upper end, extending well beyond the main concentration of riders.

- Statistical Threshold:
With Q1 = 27 and Q3 = 38, the interquartile range (IQR) is 11 years. The statistical upper fence (Q3 + 1.5 * IQR) sits at 54.5 years. Any data point above 55 years is mathematically flagged as an outlier.

- Realistic vs. Erroneous Values:
Ages between 55 and 75 represent genuine, active senior riders. Values exceeding 80 to 100 years—reaching an extreme maximum of 141 years (birth year 1878)—are invalid entries resulting from placeholder or dummy registration inputs.

- Recommendation:
Cap or filter user age (e.g., age <= 75) during downstream bivariate and multivariate analyses to prevent extreme values from distorting trend lines and averages.

```python
plt.figure(figsize=(8, 3))
sns.boxplot(x=gbf["age"])
plt.xlabel("Age (Years)")
plt.title("Boxplot of User Age (Outlier Detection)")
plt.show()
```

```python
outliers_77 = gbf[gbf["age"] > 77]
print(f"Observations with age > 77: {len(outliers_77):,}")
print(f"Percentage of dataset: {len(outliers_77) / len(gbf) * 100:.3f}%")

#remove ages above 77
gbf = gbf[gbf["age"] <= 77]
print(f"Remaining clean rows: {len(gbf):,}")
```

```python

plt.figure(figsize=(8, 3))
sns.boxplot(x=gbf["age"])

plt.xlabel("Age (Years)")
plt.title("Boxplot of User Age (Cleaned: Age <= 77)")
plt.show()

# Verification 
print(gbf["age"].describe(percentiles=[0.25, 0.5, 0.75, 0.95, 0.99]).round(1))
```

# <a id="5">  5 ) . What is the distribution for the trip duration in minutes? </a>

```python
## Review the statistic for trip duration in minutes
gbf["duration_min"].describe().round(2)
```

```python
# Make visualization to easy get insight
plt.figure(figsize=(10, 5))

sns.histplot(data=gbf, x="duration_min")

plt.xlabel("Trip Duration (Minutes)")
plt.ylabel("Trip Count")
plt.title("Distribution of Trip Duration (Full Dataset)")

plt.show()
```

```python
# Trim duration at 120 minutes (2 hours)
gbf = gbf[gbf["duration_min"] <= 120]

print(f"Remaining clean rows: {len(gbf):,}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

plt.figure(figsize=(9, 5))

bins = 10 ** np.arange(0, np.log10(120) + 0.025, 0.025)

plt.hist(
    data=gbf,
    x="duration_min",
    bins=bins,
    edgecolor="white",
    linewidth=0.3,
)
plt.xscale("log")

ticks = [1, 2, 5, 10, 15, 20, 30, 45, 60, 90, 120]
plt.xticks(ticks, [str(t) for t in ticks])

plt.xlabel("Trip Duration (Minutes - Log Scale)")
plt.ylabel("Trip Count")
plt.title("Distribution of Trip Duration (Cleaned: 1 to 120 Minutes)")

plt.show()
```

> ### <mark>Your analsys and exploration here </mark>

Distribution Shape
On a linear scale, trip duration is heavily right-skewed with an extreme upper tail. When transformed to a logarithmic scale, it reveals a clean unimodal, approximately log-normal distribution with a clear peak between 8 and 10 minutes.

Central Tendency and Commuter Patterns

Median Duration: Approximately 8.5 minutes.

Interquartile Range: 25% of trips take under 5.4 minutes, while 75% take under 13.1 minutes.

Transit Function: More than 75% of all rides finish in under 14 minutes, confirming that Ford GoBike functions primarily as a fast "last-mile" connection between public transit terminals, workplaces, and universities rather than a leisure touring service.

Outlier Analysis and Justification

Extreme Values: Unfiltered trip durations stretch up to nearly 24 hours (~1,424 minutes).

Operational Cause: Trips exceeding 120 minutes (2 hours) account for under 0.3% of total volume. Given the service's steep pricing penalties after 30 to 45 minutes, these long rides represent docking errors, forgotten returns, or day passes.

Cleaning Action: Trimming or filtering duration at 120 minutes eliminates extreme leverage points while preserving over 99.7% of genuine riding data.
