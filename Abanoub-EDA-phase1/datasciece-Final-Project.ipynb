> ### <mark>Your analsys and exploration here </mark>
Before cleaning (Full Dataset):
The raw trip duration data is highly right-skewed. The vast majority of trips cluster near zero minutes, but a long tail of extreme outliers stretches out to ~1400 minutes (over 23 hours). This heavy skew makes standard summary statistics like the mean misleading, since the extreme values pull it upward — median or a trimmed dataset gives a more honest picture.

After cleaning (1–120 minutes, log scale):
After filtering out trips longer than 120 minutes, the true shape of the distribution becomes much clearer — it approximates a log-normal distribution:

The mode sits around 8–10 minutes, indicating most trips are short to moderate in length
The distribution appears roughly symmetric on the log scale, confirming that a log transformation was the right choice to correct the original skew
A modest tail remains toward the higher end (30–120 minutes), but it's far less extreme than in the raw data
Insight
180,897 rows remained after cleaning, meaning the filter removed only a relatively small fraction of the original dataset without significantly shrinking it
Short trips (under 20 minutes) make up the large majority — consistent with typical intra-city taxi/delivery trip behavior
Durations above 120 minutes were most likely data entry errors or anomalous cases (stalled trips, GPS/logging errors, etc.) rather than genuine ride behavior, which justifies excluding them from further modeling

# <a id="10 "> 10 ) . How long is the duration of the trip for each user_type in minutes?  </a>

> ### <mark>Your analsys and exploration here </mark>

## Analysis & Exploration

The boxplot confirms and deepens the earlier finding (Q8) that **Customers take longer trips than Subscribers**, and shows how the *full distribution* differs, not just the average:

| Metric | Customer | Subscriber |
|---|---|---|
| Median | ~13 min | ~8 min |
| IQR (25th–75th percentile) | ~8.5 – 20 min | ~5.5 – 12.5 min |
| Upper whisker (max, excl. outliers) | ~38.5 min | ~23.5 min |
| Lower whisker | ~1 min | ~1 min |

### Insight
- **Customers have both a higher median and a wider spread** of trip durations — their IQR box is noticeably taller than the Subscriber box, meaning Customer trip lengths are more variable and less predictable.
- **Subscribers are far more consistent**: their trips are tightly clustered in a narrower range (roughly 5.5–12.5 minutes), consistent with routine, habitual commuting patterns where the same routes are taken repeatedly.

```python
# Make visualization to easy get insight
plt.figure(figsize=(10, 6))
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5,
)
plt.title("Correlation Matrix of All Numerical Variables")
plt.show()
```



> ### <mark>Your analsys and exploration here </mark>
## Analysis & Exploration

The correlation matrix confirms that **trip duration is essentially uncorrelated with every other numerical variable** in the dataset:

- `duration_sec`, `duration_min`, and `duration_hour` show correlations of **exactly 1.00 with each other** (as expected, since they're just unit conversions of the same underlying value) — but correlations of **near-zero (0.00 to 0.03)** with every other feature, including station location coordinates, station IDs, bike ID, member birth year, and Age.
- **No linear relationship exists** between trip duration and geographic features (start/end latitude, longitude, station IDs), meaning trip length doesn't systematically depend on *where* a trip starts or ends — a short trip is just as likely to originate from any station as a long one.
- Duration also shows **no linear correlation with `Age`/`member_birth_year`** (0.02 to -0.03), reinforcing the earlier boxplot finding: age doesn't meaningfully drive how long someone rides.
- The only strong correlations in the matrix are **structural/expected ones**: station latitude and longitude pairs (~0.99), since they're geographically linked, and `member_birth_year` vs. `Age` (**exactly -1.00**), which makes sense since Age is directly derived as `current_year - birth_year` — a perfect inverse relationship by construction, not a real-world finding.

### Insight
- The takeaway is that **trip duration behaves independently of geography, bike identity, and rider age** in a linear sense — its variation (as shown in earlier sections) is driven far more by **categorical/behavioral factors** like `user_type` (Subscriber vs. Customer) than by any numerical feature captured here.
- This also validates that **duration_sec, duration_min, and duration_hour are redundant** for modeling purposes — only one should be kept, since including more than one would introduce perfect multicollinearity.
- Similarly, **`Age` and `member_birth_year` are redundant** (perfectly inversely correlated) and only one should be retained in any predictive model to avoid multicollinearity.
- The near-zero correlations with location suggest that if duration needs to be predicted or explained further, **categorical features (user_type, gender) and possibly non-linear or interaction effects** are likely to matter more than the raw numerical station/location data.
